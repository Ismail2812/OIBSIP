```python
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

monthly_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Observation

The monthly sales trend shows considerable fluctuations in sales throughout the period from 2014 to 2017. The lowest monthly sales were recorded in February 2014, with approximately $4,520 in sales. The highest monthly sales occurred in November 2017, reaching approximately $118,448. Overall, the later years show stronger sales performance compared with the earlier period, although sales continue to vary significantly from month to month.
```

```python
quarterly_sales = df.groupby(
    df['Order Date'].dt.to_period('Q')
)['Sales'].sum()

quarterly_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby('Product Name')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
plt.figure(figsize=(12, 7))

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Best-Selling Products')
plt.xlabel('Quantity Sold')
plt.ylabel('Product Name')

plt.tight_layout()
plt.show()
```

```python
### Observation

The analysis shows that Staples is the best-selling product, with 215 units sold, followed by Staple Envelope with 170 units and Easy-staple Paper with 150 units. The remaining products in the top 10 have sales quantities ranging from 62 to 86 units. This indicates strong customer demand for stationery and office-supply products, particularly staple-related products.
```

```python
### Revenue by Product Category

This analysis compares total sales revenue across different product categories.
```

```python
category_revenue = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

category_revenue
```

```python
plt.figure(figsize=(10, 6))

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
### Observation

Technology generates the highest revenue among the three product categories, with approximately $836,154.03 in total sales. Furniture follows with approximately $741,999.80, while Office Supplies generates approximately $719,047.03. Therefore, Technology is the strongest revenue-generating category in the dataset, although all three categories contribute substantially to total sales.
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Variables')

plt.tight_layout()
plt.show()
```

```python
### Observation

The correlation analysis shows a moderate positive relationship between Sales and Profit, with a correlation coefficient of 0.479. Sales and Quantity have a weak positive relationship of 0.201. Discount has a very weak negative relationship with Sales (-0.028) but a stronger negative relationship with Profit (-0.219). This suggests that higher discount levels may be associated with lower profitability, while higher sales are generally associated with higher profit.
```

```python
## 6. Additional Visualization

### Discount vs Profit

This visualization examines whether the level of discount is associated with changes in profit.
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit'
)

plt.title('Relationship Between Discount and Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

plt.tight_layout()
plt.show()
```

```python
### Observation

The Discount versus Profit analysis shows a negative relationship between discount and profitability. The correlation between Discount and Profit is -0.219, indicating that higher discount levels tend to be associated with lower profit. This suggests that excessive discounting should be carefully managed to protect profit margins.
```

```python
## 7. Conclusion and Business Recommendations

### Conclusion

The Exploratory Data Analysis of the Superstore retail sales dataset revealed important patterns in sales performance, product demand, category revenue, and profitability. The monthly sales analysis showed considerable fluctuations over time, with stronger sales performance in the later years. Product analysis identified Staples as the best-selling product by quantity, while Technology was the highest-revenue product category. The correlation analysis showed a moderate positive relationship between Sales and Profit. In addition, the negative relationship between Discount and Profit suggests that excessive discounting may negatively affect profitability.

### Business Recommendations

1. **Prioritize high-demand products:** The company should maintain sufficient inventory of top-selling products such as Staples, Staple Envelope, and Easy-staple Paper to reduce the risk of stock shortages.

2. **Invest in high-revenue categories:** Technology generates the highest revenue, so the company should consider increasing marketing and promotional efforts for successful Technology products while maintaining strong performance in Furniture and Office Supplies.

3. **Control discount levels:** Since Discount has a negative relationship with Profit, the company should avoid excessive discounting and use targeted discounts that increase sales without significantly reducing profit margins.
```
